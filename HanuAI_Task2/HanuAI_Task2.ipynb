```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("Hanuai_task2_assignment.csv")

print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

display(df.head())
```

```python
# 1. Check data types and dataset information
print("DATASET INFORMATION")
df.info()

# 2. Check missing values
print("\nMISSING VALUES")
print(df.isnull().sum())

print("\nTotal missing cells:", df.isnull().sum().sum())

# 3. Check duplicate rows
print("\nDUPLICATE ROWS")
print("Number of duplicate rows:", df.duplicated().sum())

# 4. Check duplicate event IDs
print("\nDUPLICATE EVENT IDs")
print("Number of duplicate Event IDs:", df["Event id"].duplicated().sum())

# 5. Check summary statistics
print("\nSUMMARY STATISTICS")
display(df.describe(include="all").T)
```

```python
# Create a working copy so the original data remains unchanged
clean_df = df.copy()

# Convert date columns into datetime format
date_columns = ["Opened date", "BUILD_DATE", "IN_USE_DATE"]

for col in date_columns:
    clean_df[col] = pd.to_datetime(clean_df[col], errors="coerce")

# Clean whitespace in text columns without replacing missing values
text_columns = clean_df.select_dtypes(include="object").columns

for col in text_columns:
    clean_df[col] = clean_df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Check data quality after cleaning
print("DATASET SHAPE:", clean_df.shape)

print("\nMISSING VALUES AFTER CLEANING:")
print(clean_df.isnull().sum()[clean_df.isnull().sum() > 0])

print("\nDATE COLUMN TYPES:")
print(clean_df[date_columns].dtypes)

print("\nDUPLICATE ROWS:", clean_df.duplicated().sum())

print("\nSAMPLE CLEANED DATA:")
display(clean_df.head())
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Complaint categories
print("TOP 10 COMPLAINT CATEGORIES")
display(clean_df["COMPLAINT_CD_DESC"].value_counts().head(10))

clean_df["COMPLAINT_CD_DESC"].value_counts().head(10).plot(
    kind="bar", figsize=(10, 5), title="Top Complaint Categories"
)
plt.xlabel("Complaint Category")
plt.ylabel("Number of Events")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 2. Vehicle makes
print("COMPLAINTS BY VEHICLE MAKE")
display(clean_df["MAKE"].value_counts())

clean_df["MAKE"].value_counts().plot(
    kind="bar", figsize=(8, 5), title="Complaints by Vehicle Make"
)
plt.xlabel("Vehicle Make")
plt.ylabel("Number of Events")
plt.tight_layout()
plt.show()

# 3. Vehicle models
print("TOP 10 VEHICLE MODELS")
display(clean_df["MODEL"].value_counts().head(10))

clean_df["MODEL"].value_counts().head(10).plot(
    kind="bar", figsize=(10, 5), title="Top Models by Complaint Count"
)
plt.xlabel("Vehicle Model")
plt.ylabel("Number of Events")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 4. Manufacturing plants
print("COMPLAINTS BY BUILD PLANT")
display(clean_df["BUILD_PLANT_DESC"].value_counts())

clean_df["BUILD_PLANT_DESC"].value_counts().plot(
    kind="bar", figsize=(10, 5), title="Complaints by Build Plant"
)
plt.xlabel("Build Plant")
plt.ylabel("Number of Events")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# 5. Complaints over time
monthly_complaints = (
    clean_df.dropna(subset=["Opened date"])
    .set_index("Opened date")
    .resample("MS")
    .size()
)

print("MONTHLY COMPLAINT COUNTS")
display(monthly_complaints.to_frame("Complaint Count"))

monthly_complaints.plot(
    kind="line", marker="o", figsize=(10, 5),
    title="Complaints Over Time"
)
plt.xlabel("Month")
plt.ylabel("Number of Events")
plt.tight_layout()
plt.show()
```

```python
import re

# Combine the three text fields for analysis
text_cols = [
    "CUSTOMER_VERBATIM",
    "CAUSAL_VERBATIM",
    "CORRECTION_VERBATIM"
]

clean_df["Combined_Text"] = clean_df[text_cols].fillna("").agg(" ".join, axis=1)
clean_df["Combined_Text"] = clean_df["Combined_Text"].str.lower()

# Define keyword-based tags
tag_rules = {
    "Display/Black Screen": [
        "black screen", "blank screen", "screen is black",
        "display blank", "screen goes black"
    ],
    "No Audio/No Sound": [
        "no sound", "no audio", "no volume",
        "audio inoperative", "radio has no sound"
    ],
    "Radio/Inoperative": [
        "radio inoperative", "radio inoperable",
        "radio not working", "radio will not work"
    ],
    "Connectivity/Communication": [
        "communication", "connection", "disconnect",
        "bluetooth", "pairing", "usb", "phone connect"
    ],
    "Navigation/GPS": [
        "navigation", "gps", "satellite signal",
        "lost satellite", "map"
    ],
    "Intermittent Failure": [
        "intermittent", "sometimes", "occasionally",
        "comes and goes", "on and off"
    ],
    "Electrical/Power": [
        "power loss", "no power", "shorted",
        "electrical", "voltage", "ground"
    ],
    "Software/Programming": [
        "software", "programmed", "programming",
        "update", "reprogram", "calibration"
    ],
    "Component Replacement": [
        "replaced", "replacement", "install new"
    ],
    "Unable to Duplicate": [
        "unable to duplicate", "could not duplicate",
        "cannot duplicate", "could not verify"
    ]
}

# Extract tags for each record
def extract_tags(text):
    tags = []
    for tag, keywords in tag_rules.items():
        if any(keyword in text for keyword in keywords):
            tags.append(tag)
    return tags

clean_df["Extracted_Tags"] = clean_df["Combined_Text"].apply(extract_tags)

# Create a separate table with one row per event and tag
tag_rows = []

for _, row in clean_df.iterrows():
    for tag in row["Extracted_Tags"]:
        tag_rows.append({
            "Event id": row["Event id"],
            "Opened date": row["Opened date"],
            "MAKE": row["MAKE"],
            "MODEL": row["MODEL"],
            "COMPLAINT_CD_DESC": row["COMPLAINT_CD_DESC"],
            "Extracted Tag": tag
        })

tags_df = pd.DataFrame(tag_rows)

print("Events with at least one extracted tag:",
      clean_df["Extracted_Tags"].apply(bool).sum())

print("\nTOP EXTRACTED TAGS:")
if not tags_df.empty:
    display(tags_df["Extracted Tag"].value_counts().to_frame("Count"))
else:
    print("No keyword matches found. Review the text and keyword rules.")

print("\nSAMPLE EXTRACTED TAGS:")
display(tags_df.head(15))

# Save tag results
tags_df.to_csv("HanuAI_Task2_Extracted_Tags.csv", index=False)
print("\nSaved: HanuAI_Task2_Extracted_Tags.csv")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# Prepare text for clustering
texts = clean_df["Combined_Text"].fillna("").astype(str)

# Convert text into numerical features
vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=1000,
    ngram_range=(1, 2),
    min_df=2
)

X = vectorizer.fit_transform(texts)

# Group records into 5 clusters
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
clean_df["Text_Cluster"] = kmeans.fit_predict(X)

# Display the most important words in each cluster
terms = vectorizer.get_feature_names_out()
order = kmeans.cluster_centers_.argsort()[:, ::-1]

print("TOP TERMS IN EACH CLUSTER")

for cluster_id in range(5):
    top_terms = [terms[i] for i in order[cluster_id, :10]]
    print(f"\nCluster {cluster_id}: {', '.join(top_terms)}")

print("\nRECORDS PER CLUSTER")
display(clean_df["Text_Cluster"].value_counts().sort_index().to_frame("Record Count"))

# Save cluster assignments
clean_df[["Event id", "COMPLAINT_CD_DESC", "Text_Cluster"]].to_csv(
    "HanuAI_Task2_Cluster_Results.csv", index=False
)

print("\nSaved: HanuAI_Task2_Cluster_Results.csv")
```
