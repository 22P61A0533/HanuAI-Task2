```python
# HanuAI – ML Assessment Task 1
# Step 1: Import libraries and load the customer review dataset

import pandas as pd
from pathlib import Path

# Define the project paths
PROJECT_DIR = Path.cwd()
DATA_FILE = PROJECT_DIR / "data" / "hanuaI_reviews.xlsx"

# Load the dataset
df = pd.read_excel(DATA_FILE)

# Display basic dataset information
print("Dataset loaded successfully!")
print("Total reviews:", len(df))
print("Column names:", df.columns.tolist())

# Display the first five reviews
df.head()
```

```python
# Step 2: Validate the customer review dataset

print("Total reviews:", len(df))

# Check for missing values
print("\nMissing values in each column:")
print(df.isnull().sum())

# Check for duplicate review IDs
print("\nDuplicate IDs:", df["ID"].duplicated().sum())

# Check for invalid ratings
print(
    "Invalid ratings:",
    ((df["Rating"] < 1) | (df["Rating"] > 5) | df["Rating"].isnull()).sum()
)

# Check for empty review texts
print(
    "Empty review texts:",
    (df["Review Text"].isna() | df["Review Text"].astype(str).str.strip().eq("")).sum()
)

# Check for duplicate review texts
print(
    "Duplicate review texts:",
    df["Review Text"].duplicated().sum()
)
```

```python
# Step 3: Perform sentiment analysis using VADER

from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Initialize the sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Calculate sentiment scores for each review
df["Sentiment Score"] = df["Review Text"].apply(
    lambda text: analyzer.polarity_scores(str(text))["compound"]
)

# Classify each review based on its compound score
def classify_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

df["Sentiment"] = df["Sentiment Score"].apply(classify_sentiment)

# Display the sentiment distribution
print("Sentiment Distribution:")
print(df["Sentiment"].value_counts())

# Display a sample of the results
df[["ID", "Review title", "Rating", "Sentiment Score", "Sentiment"]].head(10)
```

```python
# Step 4: Save sentiment analysis results

from pathlib import Path

# Create the reports folder if it does not exist
REPORTS_DIR = Path.cwd() / "reports"
REPORTS_DIR.mkdir(exist_ok=True)

# Save results to Excel
output_file = REPORTS_DIR / "sentiment_results.xlsx"
df.to_excel(output_file, index=False)

print("Sentiment results saved successfully!")
print("File location:", output_file)
```

```python
# Step 5: Analyze customer feedback themes

import re
from collections import Counter

# Define the customer feedback themes and their keywords
themes = {
    "Sound Quality": ["sound", "audio", "bass", "volume", "quality"],
    "Comfort and Fit": ["comfortable", "comfort", "fit", "ears", "head"],
    "Noise Cancellation": ["noise cancellation", "anc", "noise cancelling"],
    "Battery Life": ["battery", "hours", "life", "lasting"],
    "Charging": ["charge", "charging", "charger"],
    "Microphone and Calls": ["microphone", "mic", "call", "voice"],
    "Build Quality and Durability": ["build", "quality", "crack", "broken", "durability"],
    "Price and Value": ["price", "budget", "affordable", "value", "worth"],
    "Appearance and Color": ["color", "colour", "design", "look", "appearance"],
    "Connectivity": ["bluetooth", "connect", "connection", "pairing"]
}

# Find reviews mentioning each theme
theme_results = []

for theme, keywords in themes.items():
    matching_reviews = df[
        df["Review Text"].str.lower().apply(
            lambda text: any(keyword in text for keyword in keywords)
        )
    ]

    theme_results.append({
        "Theme": theme,
        "Reviews Mentioning Theme": len(matching_reviews),
        "Positive": (matching_reviews["Sentiment"] == "Positive").sum(),
        "Negative": (matching_reviews["Sentiment"] == "Negative").sum(),
        "Neutral": (matching_reviews["Sentiment"] == "Neutral").sum()
    })

theme_df = pd.DataFrame(theme_results)

# Display the results
theme_df
```

```python
# Step 6: Verify and display customer theme counts

print(theme_df.to_string(index=False))

# Check whether sentiment counts match the number of reviews per theme
theme_df["Sentiment Count Total"] = (
    theme_df["Positive"]
    + theme_df["Negative"]
    + theme_df["Neutral"]
)

theme_df["Counts Match"] = (
    theme_df["Reviews Mentioning Theme"]
    == theme_df["Sentiment Count Total"]
)

display(theme_df)
```

```python
# Step 7: Save customer theme analysis

theme_df.to_excel(
    REPORTS_DIR / "customer_theme_analysis.xlsx",
    index=False
)

print("Customer theme analysis saved successfully!")
print("File:", REPORTS_DIR / "customer_theme_analysis.xlsx")
```

```python
# Step 8: Generate the final project summary

print("=" * 45)
print("HANUAI – ML ASSESSMENT TASK 1")
print("CUSTOMER REVIEW ANALYSIS SUMMARY")
print("=" * 45)

print("\n1. DATASET")
print("Total reviews analyzed:", len(df))
print("Missing values:", int(df.isnull().sum().sum()))

print("\n2. SENTIMENT DISTRIBUTION")
sentiment_counts = df["Sentiment"].value_counts()

for sentiment in ["Positive", "Negative", "Neutral"]:
    count = int(sentiment_counts.get(sentiment, 0))
    percentage = count / len(df) * 100
    print(f"{sentiment}: {count} reviews ({percentage:.1f}%)")

print("\n3. CUSTOMER FEEDBACK THEMES")
display(theme_df)

print("\n4. OUTPUT FILES")
print("Sentiment results: reports/sentiment_results.xlsx")
print("Theme analysis: reports/customer_theme_analysis.xlsx")

print("\nAnalysis completed.")
```

```python

# Inspect the initial-state data for review records
import re

state_script = soup.find(
    "script",
    string=lambda text: text and "window.__INITIAL_STATE__" in text
)

if state_script:
    state_text = state_script.string or state_script.get_text()
    print("Initial state found:", True)
    print("Contains reviewerName:", "reviewerName" in state_text)
    print("Contains reviewBody:", "reviewBody" in state_text)
    print("Contains reviewSource:", "reviewSource" in state_text)
    print("Contains submissionTime:", "submissionTime" in state_text)
    print("Contains review records:", '"comment"' in state_text)
else:
    print("Initial state script not found")
```

```python

import json

# Extract the JSON object assigned to window.__INITIAL_STATE__
match = re.search(
    r"window\.__INITIAL_STATE__\s*=\s*(\{.*\})\s*;?\s*$",
    state_text,
    re.DOTALL
)

if match:
    try:
        initial_state = json.loads(match.group(1))
        print("JSON parsed successfully")
        print("Top-level keys:", list(initial_state.keys()))
    except json.JSONDecodeError as e:
        print("JSON parsing failed:", e)
else:
    print("Could not isolate the initial-state JSON")
```

```python

import json

# Find the opening brace after the initial-state assignment
start = state_text.find("=", state_text.find("window.__INITIAL_STATE__")) + 1

while start < len(state_text) and state_text[start].isspace():
    start += 1

# Find the end of the JSON object while respecting strings and escapes
depth = 0
in_string = False
escaped = False
end = None

for i in range(start, len(state_text)):
    char = state_text[i]

    if in_string:
        if escaped:
            escaped = False
        elif char == "\\":
            escaped = True
        elif char == '"':
            in_string = False
    else:
        if char == '"':
            in_string = True
        elif char == "{":
            depth += 1
        elif char == "}":
            depth -= 1
            if depth == 0:
                end = i + 1
                break

if end is not None:
    initial_state = json.loads(state_text[start:end])
    print("JSON parsed successfully")
    print("Top-level keys:", list(initial_state.keys()))
else:
    print("Could not find the end of the JSON object")
```

```python
import pandas as pd
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from pathlib import Path

nltk.download("vader_lexicon", quiet=True)

# Load the validated 50-row dataset
df = pd.read_excel("data/hanuaI_reviews.xlsx")

# Analyze the review text
sia = SentimentIntensityAnalyzer()

df["Sentiment Score"] = df["Review Text"].fillna("").astype(str).apply(
    lambda text: sia.polarity_scores(text)["compound"]
)

def classify_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    return "Neutral"

df["Sentiment"] = df["Sentiment Score"].apply(classify_sentiment)

# Save detailed results
Path("reports").mkdir(exist_ok=True)

df.to_csv(
    "reports/50_review_sentiment_analysis.csv",
    index=False,
    encoding="utf-8-sig"
)
df.to_excel(
    "reports/50_review_sentiment_analysis.xlsx",
    index=False
)

# Print the analysis summary
print("Reviews analyzed:", len(df))
print("\nSentiment distribution:")
print(df["Sentiment"].value_counts())

print("\nAverage star rating:", round(df["Rating"].mean(), 2), "/ 5")

print("\nSentiment by rating:")
print(pd.crosstab(df["Rating"], df["Sentiment"]))

print("\nSource counts:")
print(df["Source"].value_counts())

print("\nSaved 50-review sentiment results.")
```

```python

# HANUAI TASK 1 — BEST BUY REVIEW SCRAPER
# Product: JLab JBuds LUX Over-Ear Headphones
# Run this entire cell in your existing Jupyter Notebook.

import requests
import pandas as pd
import json
import re
import time
from pathlib import Path
from datetime import datetime
from urllib.parse import urlparse

PRODUCT_URL = (
    "https://www.bestbuy.ca/en-ca/product/"
    "jlab-jbuds-lux-over-ear-noise-cancelling-true-wireless-headphones-mauve-only-at-best-buy/18145791"
)
PRODUCT_ID = "18145791"

OUTPUT_DIR = Path.cwd() / "data"
OUTPUT_DIR.mkdir(exist_ok=True)

session = requests.Session()
session.headers.update({
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/130.0.0.0 Safari/537.36"
    ),
    "Accept": "application/json, text/plain, */*",
    "Accept-Language": "en-CA,en;q=0.9",
    "Referer": PRODUCT_URL,
})

# Recursively find review records in a JSON response.
def find_review_lists(obj):
    found = []

    if isinstance(obj, dict):
        values = {str(k).lower(): v for k, v in obj.items()}

        has_rating = any(
            k in values for k in
            ["rating", "score", "ratingvalue", "overallrating"]
        )
        has_text = any(
            k in values for k in
            ["reviewtext", "reviewbody", "comments", "text", "body"]
        )

        if has_rating and has_text:
            found.append(obj)

        for value in obj.values():
            found.extend(find_review_lists(value))

    elif isinstance(obj, list):
        for value in obj:
            found.extend(find_review_lists(value))

    return found


def get_value(record, names):
    lowered = {str(k).lower(): v for k, v in record.items()}
    for name in names:
        value = lowered.get(name.lower())
        if value is not None and str(value).strip():
            return value
    return ""


def normalize_review(review):
    review_id = get_value(
        review, ["id", "reviewId", "review_id", "reviewID"]
    )
    title = get_value(
        review, ["title", "reviewTitle", "review_title", "headline"]
    )
    text = get_value(
        review, ["reviewText", "reviewBody", "comments", "text", "body"]
    )
    date_value = get_value(
        review, ["date", "reviewDate", "submissionTime", "createdAt"]
    )
    rating_value = get_value(
        review, ["rating", "score", "ratingValue", "overallRating"]
    )
    reviewer = get_value(
        review, ["reviewer", "reviewerName", "author", "nickname", "userName"]
    )

    try:
        rating = float(rating_value)
        if not 1 <= rating <= 5:
            return None
    except (ValueError, TypeError):
        return None

    # Normalize dates when possible.
    date = ""
    if date_value:
        try:
            date = pd.to_datetime(
                date_value, errors="raise", utc=True
            ).strftime("%Y-%m-%d")
        except (ValueError, TypeError):
            date = str(date_value)

    text = re.sub(r"\s+", " ", str(text)).strip()
    title = re.sub(r"\s+", " ", str(title)).strip()

    if not text:
        return None

    return {
        "ID": str(review_id).strip(),
        "Review title": title,
        "Review text": text,
        "Date": date,
        "Rating": rating,
        "Reviewer": str(reviewer).strip(),
        "Source": "Best Buy Canada",
        "Product URL": PRODUCT_URL,
    }


reviews_by_id = {}
attempt_log = []

# Try the documented Best Buy Canada product-review endpoint.
endpoint = (
    "https://www.bestbuy.ca/"
    f"pdp/reviewDetails/{PRODUCT_ID}"
)

print("Attempting to retrieve genuine Best Buy reviews...")

for page in range(1, 11):
    params = {
        "lang": "en",
        "source": "can",
        "page": page,
        "pageSize": 20,
    }

    try:
        response = session.get(
            endpoint, params=params, timeout=20
        )

        attempt_log.append({
            "Page requested": page,
            "HTTP status": response.status_code,
            "URL": response.url,
        })

        print(
            f"Page {page}: HTTP {response.status_code}"
        )

        if response.status_code in (401, 403, 429):
            print(
                "The server denied or rate-limited the request. "
                "Stopping without bypassing the restriction."
            )
            break

        response.raise_for_status()

        try:
            payload = response.json()
        except ValueError:
            print("The endpoint did not return JSON.")
            break

        page_reviews = find_review_lists(payload)

        if not page_reviews:
            print(
                "No review records recognized in this response. "
                "Saving the response structure for inspection."
            )
            with open(
                OUTPUT_DIR / "bestbuy_endpoint_response_sample.json",
                "w",
                encoding="utf-8",
            ) as f:
                json.dump(payload, f, indent=2, ensure_ascii=False)
            break

        new_records = 0

        for raw_review in page_reviews:
            item = normalize_review(raw_review)
            if item is None:
                continue

            # Use review ID where available; otherwise use text + rating.
            key = item["ID"] or (
                item["Review text"][:250] + "|" + str(item["Rating"])
            )

            if key not in reviews_by_id:
                reviews_by_id[key] = item
                new_records += 1

        print(
            f"Recognized {len(page_reviews)} review objects; "
            f"{new_records} new valid records; "
            f"{len(reviews_by_id)} unique records total."
        )

        if new_records == 0:
            print("No new records on this page; stopping pagination.")
            break

        time.sleep(1)

    except requests.RequestException as exc:
        print(f"Request failed on page {page}: {exc}")
        break

# Also inspect the ordinary product page's embedded public state.
if len(reviews_by_id) < 50:
    print("\nTrying the normal product page for embedded review data...")

    try:
        page_response = session.get(PRODUCT_URL, timeout=20)
        print("Product page HTTP status:", page_response.status_code)

        if page_response.ok:
            html = page_response.text

            # Extract review-like objects from common embedded JSON scripts.
            for match in re.finditer(
                r'<script[^>]*type=["\']application/json["\'][^>]*>'
                r"(.*?)</script>",
                html,
                flags=re.I | re.S,
            ):
                try:
                    payload = json.loads(match.group(1))
                except (ValueError, TypeError):
                    continue

                for raw_review in find_review_lists(payload):
                    item = normalize_review(raw_review)
                    if item is None:
                        continue

                    key = item["ID"] or (
                        item["Review text"][:250]
                        + "|" + str(item["Rating"])
                    )
                    reviews_by_id.setdefault(key, item)

    except requests.RequestException as exc:
        print("Product page request failed:", exc)

# Build the final dataframe.
df = pd.DataFrame(list(reviews_by_id.values()))

if not df.empty:
    df = df.drop_duplicates(
        subset=["ID"] if df["ID"].astype(str).str.len().gt(0).all()
        else ["Review text", "Rating"],
        keep="first",
    )

    df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
    df = df[df["Rating"].between(1, 5)]
    df = df[df["Review text"].astype(str).str.strip().ne("")]
    df = df.reset_index(drop=True)

    csv_path = OUTPUT_DIR / "bestbuy_verified_reviews.csv"
    xlsx_path = OUTPUT_DIR / "bestbuy_verified_reviews.xlsx"
    df.to_csv(csv_path, index=False, encoding="utf-8-sig")
    df.to_excel(xlsx_path, index=False)

    print("\nRESULT")
    print("Unique review records collected:", len(df))
    print("CSV saved to:", csv_path)
    print("Excel saved to:", xlsx_path)

    # Create locally sorted versions of the reviews collected.
    # These sorts do not claim to have activated Best Buy's own UI filters.
    sort_options = {
        "Most Helpful": None,  # Requires helpful-vote data.
        "Newest": "Date",
        "Highest Rating": "Rating",
        "Lowest Rating": "Rating",
        "Most Relevant": None,  # Original endpoint order retained.
    }

    sort_dir = OUTPUT_DIR / "sorted_reviews"
    sort_dir.mkdir(exist_ok=True)

    for sort_name, column in sort_options.items():
        if sort_name == "Most Helpful":
            print(
                "Most Helpful: not sorted because helpful-vote counts "
                "were not confirmed in the extracted fields."
            )
            continue

        if sort_name == "Most Relevant":
            sorted_df = df.copy()
        elif sort_name == "Newest":
            sorted_df = df.copy()
            sorted_df["_date"] = pd.to_datetime(
                sorted_df["Date"], errors="coerce"
            )
            sorted_df = sorted_df.sort_values(
                "_date", ascending=False, na_position="last"
            ).drop(columns="_date")
        elif sort_name == "Highest Rating":
            sorted_df = df.sort_values(
                "Rating", ascending=False
            )
        else:
            sorted_df = df.sort_values(
                "Rating", ascending=True
            )

        filename = re.sub(
            r"[^a-z0-9]+", "_", sort_name.lower()
        ).strip("_")
        sorted_df.to_csv(
            sort_dir / f"{filename}.csv",
            index=False,
            encoding="utf-8-sig",
        )

    pd.DataFrame(attempt_log).to_csv(
        OUTPUT_DIR / "scraping_attempt_log.csv", index=False
    )

    if len(df) >= 50:
        print(
            "\n50-record threshold reached. Check the exported records "
            "and confirm IDs, dates, reviewer names, and review text "
            "before treating the dataset as verified."
        )
    else:
        print(
            f"\nOnly {len(df)} usable records were obtained. "
            "The 50-review requirement is NOT yet met."
        )
else:
    print(
        "\nNo usable review records were extracted. "
        "No synthetic reviews were created."
    )
    print(
        "Check the HTTP status above and the saved response sample, "
        "if one was created."
    )
```

```python
import json
from pathlib import Path
import pandas as pd

html = driver.page_source
marker = '"customerReviews"'
pos = html.find(marker)

if pos == -1:
    raise ValueError("Review data not found in the current page.")

start = html.find("{", pos)
depth = 0
in_string = False
escaped = False
end = None

for i in range(start, len(html)):
    ch = html[i]
    if in_string:
        if escaped:
            escaped = False
        elif ch == "\\":
            escaped = True
        elif ch == '"':
            in_string = False
    else:
        if ch == '"':
            in_string = True
        elif ch == "{":
            depth += 1
        elif ch == "}":
            depth -= 1
            if depth == 0:
                end = i + 1
                break

if end is None:
    raise ValueError("Could not parse the embedded review data.")

obj = json.loads(html[start:end])

# Find lists of review dictionaries in the embedded object.
candidates = []

def walk(value):
    if isinstance(value, dict):
        for key, child in value.items():
            if (
                isinstance(child, list)
                and child
                and isinstance(child[0], dict)
                and any(
                    field in child[0]
                    for field in ["comment", "reviewerName", "submissionTime"]
                )
            ):
                candidates.append(child)
            walk(child)
    elif isinstance(value, list):
        for child in value:
            walk(child)

walk(obj)

if not candidates:
    print("Top-level keys:", list(obj.keys()))
    raise ValueError("No review list found. Send me the top-level keys.")

reviews = max(candidates, key=len)

rows = []
for r in reviews:
    rows.append({
        "ID": r.get("id"),
        "Review title": r.get("title"),
        "Review text": r.get("comment"),
        "Date": r.get("submissionTime"),
        "Rating": r.get("rating"),
        "Reviewer name": r.get("reviewerName"),
        "Source": r.get("reviewerLocation") or r.get("reviewSource"),
        "Verified purchaser": r.get("isVerifiedPurchaser"),
    })

df = pd.DataFrame(rows)
df["Date"] = pd.to_datetime(df["Date"], errors="coerce").dt.strftime("%Y-%m-%d")
df["Rating"] = pd.to_numeric(df["Rating"], errors="coerce")
df = df.drop_duplicates(subset=["ID"], keep="first")
df = df[df["Review text"].fillna("").str.strip().ne("")]

out = Path.cwd() / "data"
out.mkdir(parents=True, exist_ok=True)

df.to_csv(out / "bestbuy_embedded_reviews.csv", index=False, encoding="utf-8-sig")
df.to_excel(out / "bestbuy_embedded_reviews.xlsx", index=False)

print("Extracted reviews:", len(df))
print("\nSource counts:")
print(df["Source"].value_counts(dropna=False).to_string())
print("\nRating counts:")
print(df["Rating"].value_counts(dropna=False).sort_index().to_string())
print("\nSample records:")
display(df.head(5))
```
